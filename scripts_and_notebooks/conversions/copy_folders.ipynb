```python
import os
import shutil
import glob
from tqdm import tqdm

class_mapping = {0: "CFCBK", 1: "FCBK", 2: "Zigzag"}
image_width = 640
image_height = 640
```

```python
def copy_files(input_folder, output_folder):
    if not os.path.exists(input_folder):
        raise FileNotFoundError(f"Input folder '{input_folder}' not found")
    os.makedirs(output_folder, exist_ok=True)
    files = [f for f in os.listdir(input_folder) if f.endswith(".txt")]
    files = sorted(files)
    for file_name in tqdm(files, desc="Copying Files", unit="file"):
        input_path = os.path.join(input_folder, file_name)
        output_path = os.path.join(output_folder, file_name)
        shutil.copy(input_path, output_path)
```

```python
input_folders = [
    "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/region_performance/test_bihar_same_class_count_10_120_1000/labels",
    "/home/patel_zeel/kiln_compass_24/data/m0_obb_without_empty_val/labels",
    "/home/rishabh.mondal/Brick-Kilns-project/ijcai_2025_kilns/data/swinir_data/test_bihar_same_class_count_10_120_1000_4x/labels",
]

output_folders = [
    "../../data/bihar/val/labels",
    "../../data/m0/val/labels",
    "../../data/swinir/test_bihar_same_class_count_10_120_1000_4x/labels",
]

for input_folder, output_folder in zip(input_folders, output_folders):
    try:
        copy_files(input_folder, output_folder)
        print(f"Copied files from {input_folder} to {output_folder}")
        print(len(os.listdir(input_folder)), "files found")
        print(len(os.listdir(output_folder)), "files copied")
    except Exception as e:
        print(f"Error copying files from {input_folder} to {output_folder}: {e}")
```
